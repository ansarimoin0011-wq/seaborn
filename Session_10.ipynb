```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#                                                              Task 1
data = {
    "day": [
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday"
    ],
    
    "songs_streamed": [
        120,
        150,
        135,
        180,
        220,
        300,
        260
    ]
}
df = pd.DataFrame(data)
```

```python

sns.barplot(df,x="day",y="songs_streamed")
```

```python
sns.set(context="poster") # Actual graph mein bars same data represent karenge; presentation elements larger honge.
#paper    → small
#notebook  → normal
#talk      → large
#poster    → very large
sns.barplot(df,x="day",y="songs_streamed")
```

```python
#                                                          Task -2
data = {
    "cuisine": [
        "Indian", "Indian", "Indian", "Indian", "Indian",
        "Chinese", "Chinese", "Chinese", "Chinese",
        "Italian", "Italian", "Italian",
        "Indian", "Chinese", "Italian",
        "Indian", "Chinese", "Indian", "Italian", "Chinese"
    ]
}

df = pd.DataFrame(data)
```

```python
#                                                                Task 3
data =({
    "Day": [
        'Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5', 'Day 6', 'Day 7',
        'Day 8', 'Day 9', 'Day 10', 'Day 11', 'Day 12', 'Day 13', 'Day 14'
    ],
    "Steps": [
        5200, 6800, 7500, 6100, 8200, 9000, 7800,
        6500, 7200, 8500, 9300, 8800, 7600, 10000
    ]
})
df = pd.DataFrame(data)
```

```python
sns.set_context("notebook")  # note book style page.
sns.lineplot(df,x="Day",y="Steps",color=sns.color_palette("muted")[0])
# [ color_palette ] ka benefit tab zyada hota hai jab multiple groups/hue hon.
```
