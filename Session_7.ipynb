```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
#                                                                  Task-1
data = {
    "delivery_time": [30, 45, 25, 50, 35, 40, 28, 55, 32, 48,
                      38, 42, 27, 60, 36, 44, 31, 52, 29, 46],

    "order_value": [500, 800, 350, 1000, 600, 750, 400, 1200, 550, 900,
                    650, 700, 300, 1300, 450, 850, 520, 1100, 380, 950],

    "tip_amount": [50, 80, 30, 100, 60, 70, 40, 120, 55, 90,
                   65, 75, 25, 130, 45, 85, 50, 110, 35, 95],

    "rating": [4.5, 4.0, 4.8, 3.5, 4.2, 4.0, 4.7, 3.2, 4.5, 3.8,
               4.1, 4.0, 4.9, 3.0, 4.3, 3.9, 4.6, 3.4, 4.8, 3.7],

    "delivery_distance": [2, 5, 1, 7, 3, 4, 2, 8, 3, 6,
                          4, 5, 1, 9, 2, 6, 3, 7, 2, 5]
}

df = pd.DataFrame(data)
```

```python
correlation = df.corr()  # .corr(function hai )
print(correlation)
```

```python
#                                                                   Task-2
sns.heatmap(correlation, annot=True) 
#annot=True -> values boxes ke andar dikhengi
#  Us correlation ko heatmap mein show karta hai.
```

```python
#                                                                  Task-3
sns.heatmap(correlation, annot=True, cmap="coolwarm")
# 0.20
# 0.65
# 0.85
#Inmein -0.85 strongest negative correlation hai.
#Important: Negative correlation mein 0 se sabse door negative value strongest hoti hai.
#annot=True -> values boxes ke andar dikhengi
#cmap="coolwarm" -> heatmap ka color change karega
```
