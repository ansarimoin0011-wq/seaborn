```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
#                                                                     Task - 1
df = {
    'Food': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'Orders': [120, 90, 150, 110, 80]
}
```

```python
sns.barplot(df,x="Food", y="Orders")
plt.savefig("food_order", dpi=300)  # food_oder ke naam se 300DPI pr "PNG" save ki hai
```

```python
 #                                                               Task-5
data= {
    "order_value": [
        200, 300, 400, 500, 600,
        250, 350, 450, 550, 650 ],
   "delivery_time": [
        20, 25, 30, 35, 40,
        22, 28, 32, 38, 42],
   "rating": [
        4.8, 4.5, 4.2, 4.0, 3.8,
        4.7, 4.4, 4.1, 3.9, 3.7]}
df = pd.DataFrame(data)
```

```python
df
```

```python
correlation_matrix = df.corr()
correlation_matrix.round(2)
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm")
```
